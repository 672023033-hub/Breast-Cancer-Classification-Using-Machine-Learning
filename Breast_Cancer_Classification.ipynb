Tujuan dari code ini untuk membangun model klasifikasi untuk membedakan antara kanker payudara maligna dan benigna

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import missingno as msno

import warnings
warnings.filterwarnings('ignore')


plt.style.use('ggplot')
```

```python
df = pd.read_csv('data (2).csv')
df.head()
```

#Data Preprocessing


```python
df.drop(['id', 'Unnamed: 32'], axis = 1, inplace = True)

#ini untuk menghapus kolom yang tidak mempunyai nilai prediktif, yang dihapus itu data yang tidak relevan atau yang mungkin kosong
```

```python
df.diagnosis.unique()

#sesuai namanya buat ngecek diagnosa yang nilainya unik
#output M itu malignant = ganas, B itu benign = jinak
```

```python
df['diagnosis'] = df['diagnosis'].apply(lambda val: 1 if val == 'M' else 0)

# ubah teks ke angka, sesuai kodenya kalo semua value M bakalan diubah jadi 1 kalo memenuhi kondisi, selain 1(M) berarti jadi 0(B)
```

```python
df.describe()
#nampilin statik desktriptif dari data numerik
```

```python
df.info()
```

```python
# checking for null values

df.isna().sum()
```

Exploratory Data Analysis (EDA)

```python
#ini digunakan untuk memvisualisasikan distribusi dari semua fitur numerik dalam dataset kanker payudara (Breast Cancer Dataset). Visualisasi dilakukan dengan histogram dan kurva KDE (Kernel Density Estimation), satu per fitur.

plt.figure(figsize = (20, 15))
plotnumber = 1

for column in df:
    if plotnumber <= 30:
        ax = plt.subplot(5, 6, plotnumber)
        sns.distplot(df[column])
        plt.xlabel(column)

    plotnumber += 1

plt.tight_layout()
plt.show()
```

```python
# heatmapKode ini digunakan untuk membuat heatmap korelasi antar fitur, sehingga kamu bisa melihat pola hubungan antar fitur, dan memutuskan fitur mana yang paling berpengaruh terhadap diagnosis kanker, serta menghindari fitur-fitur yang terlalu mirip satu sama lain (redundan).

plt.figure(figsize = (20, 12))

corr = df.corr()
mask = np.triu(np.ones_like(corr, dtype = bool))

sns.heatmap(corr, mask = mask, linewidths = 1, annot = True, fmt = ".2f")
plt.show()
```

```python
# removing highly correlated features
#hapus fitur-fitur yang punya korelasi sangat tinggi (> 0.92) satu sama lain,
#karena fitur seperti ini cenderung membawa informasi yang redundan/berulang.

corr_matrix = df.corr().abs()

mask = np.triu(np.ones_like(corr_matrix, dtype = bool))
tri_df = corr_matrix.mask(mask)

to_drop = [x for x in tri_df.columns if any(tri_df[x] > 0.92)]

df = df.drop(to_drop, axis = 1)

print(f"The reduced dataframe has {df.shape[1]} columns.")
```

```python
# creating features and label
#Kode ini memisahkan data menjadi fitur (X) dan label (y), yang merupakan langkah wajib sebelum data dimasukkan ke dalam model klasifikasi
#seperti Decision Tree, SVM, atau Gradient Boosting.

X = df.drop('diagnosis', axis = 1)
y = df['diagnosis']
```

```python
# splitting data into training and test set
#Kode ini membagi dataset menjadi data pelatihan dan pengujian untuk membangun model dengan benar dan menghindari overfitting.
#Data pelatihan digunakan untuk melatih model, sedangkan data pengujian digunakan untuk mengukur kemampuan generalisasi model terhadap data baru.

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.30, random_state = 0)
```

```python
# scaling data
#caling ini menyiapkan fitur agar memiliki skala yang sama, meningkatkan akurasi dan efisiensi model.
#Wajib banget dilakukan terutama sebelum menggunakan algoritma seperti:
#Logistic Regression, SVM, KNN, Neural Network, Gradient Boosting

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

# Logistic Regression


```python
# fitting data to model ; Mengimpor dan melatih model Logistic Regression menggunakan data training (X_train, y_train).

from sklearn.linear_model import LogisticRegression

log_reg = LogisticRegression()
log_reg.fit(X_train, y_train)
```

```python
# model predictions ; Menghasilkan prediksi terhadap data uji (X_test).

y_pred = log_reg.predict(X_test)
```

```python
# accuracy score ; kurasi training: 0.9899 (≈99%), Akurasi testing: 0.9590 (≈96%)
#Artinya model sangat bagus dan tidak overfitting (karena akurasi train dan test masih seimbang).

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

print(accuracy_score(y_train, log_reg.predict(X_train)))

log_reg_acc = accuracy_score(y_test, log_reg.predict(X_test))
print(log_reg_acc)
```

```python
# confusion matrix ; ngecek diagona bener, 106 B benar, 58 M benar
# 2 di dikira M, 5 dikira B

print(confusion_matrix(y_test, y_pred))
```

```python
# classification report
# Akurasi tinggi: ~96%
#Hanya 5 kasus malignan yang salah klasifikasi (FN) — perlu perhatian, tapi masih tergolong bagus
#F1-score dan recall cukup tinggi

print(classification_report(y_test, y_pred))
```

# Support Vector Classifier (SVC)

```python
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV

svc = SVC()
parameters = {
    'gamma' : [0.0001, 0.001, 0.01, 0.1],
    'C' : [0.01, 0.05, 0.5, 0.1, 1, 10, 15, 20]
}

grid_search = GridSearchCV(svc, parameters)
grid_search.fit(X_train, y_train)
```

```python
# best parameters

grid_search.best_params_
```

```python
# best accuracy
#Akurasi Training: 98.74%
#Akurasi Testing: 97.66% → tertinggi dari semua model sebelumnya!

grid_search.best_score_
```

```python
svc = SVC(C = 10, gamma = 0.01)
svc.fit(X_train, y_train)
```

```python
# model predictions

y_pred = svc.predict(X_test)
```

```python
# accuracy score

print(accuracy_score(y_train, svc.predict(X_train)))

svc_acc = accuracy_score(y_test, svc.predict(X_test))
print(svc_acc)
```

```python
# confusion matrix
# Hanya 4 kesalahan total, dengan 3 false negatives (FN) — terkecil dari semua model sejauh ini
# FN itu yang salah salah diagnosa di ats

print(confusion_matrix(y_test, y_pred))
```

```python
# classification report
# F1-score untuk class 1 (malignant) = 0.97 → sangat tinggi, artinya model ini bagus banget buat deteksi kanker


print(classification_report(y_test, y_pred))
```
